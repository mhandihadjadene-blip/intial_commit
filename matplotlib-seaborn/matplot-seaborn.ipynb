```python
batsman = pd.read_csv('CampusX_sharma-kohli.csv')
```

```python
plt.title('Rohit Sharma VS Virat Kohli runs over the last seasons')
plt.xlabel('Season')
plt.ylabel('Runs Scored')

plt.plot(batsman['index'], batsman['V Kohli'], color = 'yellow', linestyle = 'dotted',
          marker = 'D', markersize=10, label='Virat')
plt.plot(batsman['index'],batsman['RG Sharma'], linestyle='solid', color ='purple', label ='Sharma', 
         marker = 'o', markersize = 3, linewidth = 2 )
plt.legend(loc ='best')
plt.grid()

plt.ylim(200,700)
```

```python
batsman = pd.read_csv('CampusX_batsman_season_record.csv')
```

```python
batsman.head()
```

```python
plt.bar(batsman['batsman'], batsman['2015'])
```

```python
#side to side
plt.xticks(np.arange(batsman.shape[0]), batsman['batsman'])
plt.xticks(rotation = 70)

plt.bar(np.arange(batsman.shape[0])-0.2, batsman['2015'],width = 0.2)
plt.bar(np.arange(batsman.shape[0]), batsman['2016'],width = 0.2)
plt.bar(np.arange(batsman.shape[0])+0.2, batsman['2017'],width = 0.2)

plt.show()
```

```python
#scacked bar chart
plt.bar(batsman['batsman'], batsman['2015'], label = '2015', color = 'blue')
plt.bar(batsman['batsman'], batsman['2016'],bottom= batsman['2015'], label = '2016', color ='green')
plt.bar(batsman['batsman'], batsman['2017'],bottom= batsman['2015']+batsman['2016'], label = '2017', color = 'red')

plt.legend(loc = 'best')
plt.show()
```
