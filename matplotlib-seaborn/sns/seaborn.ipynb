```python
import seaborn as sns
import plotly.express as px                 
import matplotlib.pyplot as plt
```

```python
tips = sns.load_dataset('tips')
tips.head()
```

```python
sns.scatterplot(tips, x ='total_bill', y='tip', hue='sex', style='time', size ='size')
```

```python
# relationplot
sns.relplot(tips, x='total_bill', y='tip', kind='scatter', hue='sex', style ='time', size ='size')
```

```python
# lineplot
gap = px.data.gapminder()
sample_gap = gap[gap['country']=='India']
sample_gap
```

```python
sns.lineplot(data = sample_gap, x='year', y='lifeExp')    
```

```python
sns.relplot(sample_gap, x='year', y='lifeExp', kind='line')
```

```python
sample_gap_country = gap[gap['country'].isin(['India','Pakistan','China'])]
```

```python
sns.relplot(kind='line', data=sample_gap_country, x='year',y='lifeExp', hue ='country')
```

```python
sample_gap_continent = gap[gap['country'].isin(['India','Colombia','France'])]
```

```python
sns.relplot(kind='line', data=sample_gap_continent, x='year',y='lifeExp', hue ='country', style ='continent', size='continent')
```

### Heatmap

```python
gap_h_country = gap.pivot(index='country',columns='year', values='lifeExp')
```

```python
plt.figure(figsize=(15,15))

sns.heatmap(gap_h_country)
```

```python
gap_europe = gap[gap['continent']=='Europe'].pivot(index='country',columns='year', values='lifeExp')
```

```python
plt.figure(figsize=(15,15))

sns.heatmap(gap_europe, annot=True, linewidths= 0.5, cmap= 'viridis')
```
