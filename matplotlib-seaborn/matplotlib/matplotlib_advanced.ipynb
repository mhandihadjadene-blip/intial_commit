```python
fig, ax = plt.subplots(figsize = (15,6))

ax.scatter(batter['avg'], batter['strike_rate'], color ='red', marker = '+')
ax.set_title('Something')
ax.set_xlabel('Avg')
ax.set_ylabel('Strike-Rate')
fig.show()
```

```python
#Side to side
fig, ax= plt.subplots(nrows= 2, ncols=1, sharex= True, figsize = (15,6))

ax[0].scatter(batter['avg'], batter['strike_rate'], color ='red')
ax[1].scatter(batter['avg'], batter['runs'])

ax[0].set_title('Avg vs Strike-Rate')
ax[0].set_ylabel('Strike-Rate')
ax[1].set_title('Avg vs Number of Runs')
ax[1].set_ylabel('Number of Runs')
```

```python
#Side to side
fig, ax= plt.subplots(nrows= 2, ncols=2, figsize = (10,10))

ax[0,0].scatter(batter['avg'], batter['strike_rate'], color ='red')
ax[0,1].scatter(batter['avg'], batter['runs'])
ax[1,0].hist(batter['strike_rate'])
ax[1,1].hist(batter['runs'])

ax[0,0].set_title('Avg vs Strike-Rate')
ax[0,0].set_ylabel('Strike-Rate')
ax[0,1].set_title('Avg vs Number of Runs')
ax[0,1].set_ylabel('Number of Runs')
```

```python
#
fig = plt.figure()

ax1 = fig.add_subplot(2,1,1)
ax1.scatter(batter['avg'], batter['strike_rate'], color ='red')
ax2 = fig.add_subplot(2,1,2)
ax2.scatter(batter['avg'], batter['runs'])
```

```python
#Line plot
x= [0,1,5,25]
y=[0,10,13,0]
z=[0,13,20,9]

fig = plt.figure()
ax= plt.subplot(projection= '3d')

ax.plot3D(x,y,z, color='red', marker= 'o',markersize = 10)
```

```python
#surface plot
x = np.linspace(-10,10,100)
y = np.linspace(-10,10,100)
```

```python
xx, yy= np.meshgrid(x,y)
xx
```

```python
z = xx**2 + yy**2
```

```python
fig = plt.figure(figsize=(12,8))
ax = plt.subplot(projection='3d')

p = ax.plot_surface(xx,yy,z, cmap='viridis')
fig.colorbar(p)
```

```python
z = np.sin(xx) + np.cos(yy)
fig = plt.figure(figsize=(12,8))
ax = plt.subplot(projection='3d')

p = ax.plot_surface(xx,yy,z, cmap='viridis')
fig.colorbar(p)
```

```python
z = np.sin(xx) + np.log(yy)
fig = plt.figure(figsize=(12,8))
ax = plt.subplot(projection='3d')

p = ax.plot_surface(xx,yy,z, cmap='viridis')
fig.colorbar(p)
```

```python
#contour plot
z = xx**2 + yy**2

fig = plt.figure(figsize=(12,8))
ax = plt.subplot()

p = ax.contour(xx,yy,z, cmap = 'viridis')
fig.colorbar(p)
```

```python
z = xx**2 + yy**2

fig = plt.figure(figsize=(12,8))
ax = plt.subplot()

p = ax.contourf(xx, yy,z, cmap ='viridis')
fig.colorbar(p)
```
